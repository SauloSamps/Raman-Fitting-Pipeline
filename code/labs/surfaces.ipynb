```python
import numpy as np
import pandas as pd
import itertools
import random
import math
from lmfit import Model
from RamanModel import *
```

```python
def generate_curves(params, output_csv="synthetic_raman_dataset.csv"):
    """
    Generate synthetic Raman curves with 2 noiseless Lorentzian peaks.

    Parameters
    ----------
    params : list of tuples
        Each tuple contains:
        (
            peak_distance,
            relative_intensity,
            relative_gamma,
            max_gamma
        )

    output_csv : str
        Output CSV filename.
    """

    x = np.linspace(0, 1, 1000)

    dataset_rows = []

    for curve_id, (
        peak_distance,
        relative_intensity,
        relative_gamma,
        max_gamma
    ) in enumerate(params):

        if curve_id % 100 == 0:
            progress = 100 * curve_id / len(params)
            print(
                f"[{progress:6.2f}%] "
                f"{curve_id}/{len(params)} curves processed",
                flush=True
            )

        # ============================================================
        # TRUE PARAMETERS
        # ============================================================

        # ----- Peak centers -----
        #center_1 = np.random.uniform(0.1, 0.9 - peak_distance)
        #center_2 = center_1 + peak_distance
        center_1 = 0.5 - (peak_distance/2)
        center_2 = 0.5 + (peak_distance/2)
        
        # ----- Peak amplitudes -----
        # One peak ALWAYS has amplitude 1
        amp_max = 1.0
        amp_min = relative_intensity

        # Randomly decide which peak is the strongest
        if np.random.rand() < 0.5:
            amplitude_1 = amp_max
            amplitude_2 = amp_min
        else:
            amplitude_1 = amp_min
            amplitude_2 = amp_max

        # ----- Peak gammas -----
        # One peak ALWAYS has gamma = max_gamma
        gamma_max = max_gamma
        gamma_min = gamma_max * relative_gamma

        # Randomly assign larger gamma
        if np.random.rand() < 0.5:
            gamma_1 = gamma_max
            gamma_2 = gamma_min
        else:
            gamma_1 = gamma_min
            gamma_2 = gamma_max

        # ============================================================
        # GENERATE CURVE
        # ============================================================

        y = (
            lorentzian(x, center_1, gamma_1, amplitude_1)
            + lorentzian(x, center_2, gamma_2, amplitude_2)
        )

        # ============================================================
        # FIT MODEL
        # ============================================================

        model = (
            Model(lorentzian, prefix='p1_')
            + Model(lorentzian, prefix='p2_')
        )

        fit_params = model.make_params()

        # Random initial guesses
        fit_params["p1_x0"].set(value=np.random.uniform(0, 1))
        fit_params["p1_gamma"].set(
            value=np.random.uniform(0.001, 1),
            min=1e-6
        )
        fit_params["p1_maximum"].set(
            value=np.random.uniform(0.01, 1),
            min=0
        )

        fit_params["p2_x0"].set(value=np.random.uniform(0, 1))
        fit_params["p2_gamma"].set(
            value=np.random.uniform(0.001, 1),
            min=1e-6
        )
        fit_params["p2_maximum"].set(
            value=np.random.uniform(0.01, 1),
            min=0
        )

        result = model.fit(y, fit_params, x=x)

        # ============================================================
        # EXTRACT FITTED PARAMETERS
        # ============================================================

        true_peaks = [
            {
                "center": center_1,
                "gamma": gamma_1,
                "amplitude": amplitude_1,
            },
            {
                "center": center_2,
                "gamma": gamma_2,
                "amplitude": amplitude_2,
            }
        ]

        fitted_peaks = [
            {
                "center": result.params["p1_x0"].value,
                "gamma": result.params["p1_gamma"].value,
                "amplitude": result.params["p1_maximum"].value,
            },
            {
                "center": result.params["p2_x0"].value,
                "gamma": result.params["p2_gamma"].value,
                "amplitude": result.params["p2_maximum"].value,
            }
        ]

        # Sort peaks by center before comparison
        true_peaks = sorted(true_peaks, key=lambda p: p["center"])
        fitted_peaks = sorted(fitted_peaks, key=lambda p: p["center"])

        # ============================================================
        # CHECK FIT SUCCESS
        # ============================================================

        failed_parameters = []

        for peak_idx, (t, f) in enumerate(zip(true_peaks, fitted_peaks), start=1):

            for key in ["center", "gamma", "amplitude"]:

                rel_error = abs((f[key] - t[key]) / t[key])

                if rel_error > 0.05:
                    failed_parameters.append(
                        f"peak_{peak_idx}_{key}"
                    )

        fit_success = int(len(failed_parameters) == 0)

        failed_parameters_str = (
            ";".join(failed_parameters)
            if failed_parameters
            else ""
        )

        # ============================================================
        # SAVE ROW
        # ============================================================

        dataset_rows.append({

            # Dataset variables
            "curve_id": curve_id,
            "peak_distance": peak_distance,
            "relative_intensity": relative_intensity,
            "relative_gamma": relative_gamma,
            "max_gamma": max_gamma,

            # Peak 1 true params
            "p1_center": true_peaks[0]["center"],
            "p1_gamma": true_peaks[0]["gamma"],
            "p1_amplitude": true_peaks[0]["amplitude"],

            # Peak 2 true params
            "p2_center": true_peaks[1]["center"],
            "p2_gamma": true_peaks[1]["gamma"],
            "p2_amplitude": true_peaks[1]["amplitude"],

            # Fit result
            "fit_success": fit_success,
            "failed_parameters": failed_parameters_str,
        })

    # ================================================================
    # SAVE CSV
    # ================================================================

    df = pd.DataFrame(dataset_rows)

    df.to_csv(output_csv, index=False)

    return df
```

```python
# ================================================================
# PARAMETER SPACE
# ================================================================

num_samples = 100000

distancias = np.linspace(0.005, 0.05, 10)
intensidade_rel = np.linspace(0.01, 1, 100)
gamma_rel = np.linspace(0.01, 1, 100)
max_gamma = np.linspace(0.1, 1, 100)

params = [
    (
        random.choice(distancias),
        random.choice(intensidade_rel),
        random.choice(gamma_rel),
        random.choice(max_gamma),
    )
    for _ in range(num_samples)
]

# ================================================================
# GENERATE DATASET
# ================================================================

df = generate_curves(params)
```

```python
import matplotlib.pyplot as plt


def plot_variable_distribution(
    df,
    variable,
    bins=50,
    figsize=(8, 5)
):
    """
    Plot histogram of a dataset variable.

    Parameters
    ----------
    df : pandas.DataFrame
        Dataset dataframe.

    variable : str
        Column name to plot.

    bins : int
        Number of histogram bins.

    figsize : tuple
        Figure size.
    """

    if variable not in df.columns:
        raise ValueError(f"Column '{variable}' not found in dataframe.")

    plt.figure(figsize=figsize)

    plt.hist(df[variable], bins=bins)

    plt.xlabel(variable)
    plt.ylabel("Count")
    plt.title(f"Distribution of {variable}")

    plt.grid(True)

    plt.show()
```

```python
# Plot dataset variable distributions

plot_variable_distribution(df, "peak_distance")
plot_variable_distribution(df, "relative_intensity")
plot_variable_distribution(df, "relative_gamma")
plot_variable_distribution(df, "max_gamma")

# Optional: success distribution

#plot_variable_distribution(df, "fit_success", bins=2)
#failed_df = df[df["fit_success"] == 0]

#plot_variable_distribution(failed_df, "peak_distance")
```

```python
import numpy as np
import pandas as pd
from lmfit import Model
from RamanModel import *

# distancias = np.linspace(0, 0.5, 0.05)
# intensidades = np.linspace(0.01, 1, 0.01)
# gamma_rel = np

# for distancia in distancias
#    for intensidade in intensidades
        

def generate_curves(num_samples, distancia, intensidade_relativa, ):
    center_range = (0.25, 0.75) # <-- ao inves disso
    amplitude_range = (0.01, 1)
    gamma_range = (0.001, 0.1)

    results_summary = []
    results_params = []

    x = np.linspace(0, 1, 1000)

    for curve_id in range(num_samples):
        y = np.zeros_like(x)
        true_params = []

        # --- Generate 2 peaks ---
        for i in range(2):
            center = np.random.uniform(*center_range)
            amplitude = np.random.uniform(*amplitude_range)
            gamma = np.random.uniform(*gamma_range)

            y += lorentzian(x, center, gamma, amplitude)

            true_params.append({
                "center": center,
                "gamma": gamma,
                "amplitude": amplitude,
            })

        # --- Sort peaks left → right ---
        true_params = sorted(true_params, key=lambda p: p["center"])

        # --- Build lmfit model ---
        model = Model(lorentzian, prefix='p1_') + Model(lorentzian, prefix='p2_')

        params = model.make_params()

        # Initial guesses (important for convergence)
        for i, p in enumerate(true_params):
            prefix = f"p{i+1}_"
            #params[f"{prefix}x0"].set(value=p["center"] * np.random.uniform(0.9, 1.1))
            #params[f"{prefix}gamma"].set(value=p["gamma"] * np.random.uniform(0.9, 1.1), min=1e-6)
            #params[f"{prefix}maximum"].set(value=p["amplitude"] * np.random.uniform(0.9, 1.1), min=0)

            params[f"{prefix}x0"].set(value=np.random.uniform(0, 1))
            params[f"{prefix}gamma"].set(value=np.random.uniform(0.001, 0.1), min=1e-6)
            params[f"{prefix}maximum"].set(value=np.random.uniform(0.01, 1), min=0)

        # --- Fit ---
        result = model.fit(y, params, x=x)

        # --- Extract fitted parameters ---
        fitted_params = []
        for i in range(2):
            prefix = f"p{i+1}_"
            fitted_params.append({
                "center": result.params[f"{prefix}x0"].value,
                "gamma": result.params[f"{prefix}gamma"].value,
                "amplitude": result.params[f"{prefix}maximum"].value,
            })

        # Sort fitted peaks too (important!)
        fitted_params = sorted(fitted_params, key=lambda p: p["center"])

        # --- Check relative error (≤5%) ---
        success = 1
        for t, f in zip(true_params, fitted_params):
            for key in ["center", "gamma", "amplitude"]:
                if t[key] == 0:
                    continue
                rel_error = abs((f[key] - t[key]) / t[key])
                if rel_error > 0.05:
                    success = 0
                    break
            if success == 0:
                break

        # --- Compute derived values ---
        amps = [p["amplitude"] for p in true_params]
        rel_intensity = min(amps) / max(amps)

        gamma1 = true_params[0]["gamma"]
        gamma2 = true_params[1]["gamma"]

        peak_distance = abs(true_params[1]["center"] - true_params[0]["center"])

        # --- Save summary row ---
        results_summary.append({
            "curve_id": curve_id,
            "relative_intensity": rel_intensity,
            "gamma_1": gamma1,
            "gamma_2": gamma2,
            "peak_distance": peak_distance,
            "fit_success": success
        })

        # --- Save detailed parameters ---
        for i, p in enumerate(true_params):
            results_params.append({
                "curve_id": curve_id,
                "peak_id": i,
                "center": p["center"],
                "gamma": p["gamma"],
                "amplitude": p["amplitude"]
            })

    # --- Write CSVs ---
    df_summary = pd.DataFrame(results_summary)
    df_params = pd.DataFrame(results_params)

    df_summary.to_csv("curve_summary.csv", index=False)
    df_params.to_csv("curve_parameters.csv", index=False)

    return df_summary, df_params
    
```

```python
%matplotlib notebook

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm

# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv("curve_summary.csv")

# ============================================================
# PARAMETERS
# ============================================================

N_DISTANCE_BINS = 5

# bins for voxel probability volume
N_GAMMA1_BINS = 8
N_GAMMA2_BINS = 8
N_INTENSITY_BINS = 8

# ============================================================
# CREATE PEAK DISTANCE BINS
# ============================================================

distance_bins = np.linspace(
    df["peak_distance"].min(),
    df["peak_distance"].max(),
    N_DISTANCE_BINS + 1
)

df["distance_bin"] = pd.cut(
    df["peak_distance"],
    bins=distance_bins,
    include_lowest=True,
    labels=False
)

# ============================================================
# FUNCTION: RAW 3D SCATTER
# ============================================================

def plot_raw_scatter(subdf, bin_id, dmin, dmax):

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    success = subdf[subdf["fit_success"] == 1]
    failure = subdf[subdf["fit_success"] == 0]

    ax.scatter(
        success["gamma_1"],
        success["gamma_2"],
        success["relative_intensity"],
        c='green',
        s=20,
        alpha=0.7,
        label='Success'
    )

    ax.scatter(
        failure["gamma_1"],
        failure["gamma_2"],
        failure["relative_intensity"],
        c='red',
        s=20,
        alpha=0.7,
        label='Failure'
    )

    ax.set_xlabel("Gamma 1")
    ax.set_ylabel("Gamma 2")
    ax.set_zlabel("Relative Intensity")

    title = (
        f"Peak Distance Range: "
        f"{dmin:.4f} ≤ Δx < {dmax:.4f}"
    )
    ax.set_title(title)

    ax.legend()

    plt.tight_layout()
    plt.show()

# ============================================================
# FUNCTION: PROBABILITY VOXEL VOLUME
# ============================================================

def plot_probability_volume(subdf, bin_id, dmin, dmax):

    # --------------------------------------------------------
    # DEFINE BINS
    # --------------------------------------------------------

    g1_bins = np.linspace(
        df["gamma_1"].min(),
        df["gamma_1"].max(),
        N_GAMMA1_BINS + 1
    )

    g2_bins = np.linspace(
        df["gamma_2"].min(),
        df["gamma_2"].max(),
        N_GAMMA2_BINS + 1
    )

    ri_bins = np.linspace(
        df["relative_intensity"].min(),
        df["relative_intensity"].max(),
        N_INTENSITY_BINS + 1
    )

    # --------------------------------------------------------
    # ASSIGN BIN INDICES
    # --------------------------------------------------------

    subdf = subdf.copy()

    subdf["g1_bin"] = pd.cut(
        subdf["gamma_1"],
        bins=g1_bins,
        labels=False,
        include_lowest=True
    )

    subdf["g2_bin"] = pd.cut(
        subdf["gamma_2"],
        bins=g2_bins,
        labels=False,
        include_lowest=True
    )

    subdf["ri_bin"] = pd.cut(
        subdf["relative_intensity"],
        bins=ri_bins,
        labels=False,
        include_lowest=True
    )

    # --------------------------------------------------------
    # COMPUTE SUCCESS PROBABILITY PER VOXEL
    # --------------------------------------------------------

    grouped = subdf.groupby(
        ["g1_bin", "g2_bin", "ri_bin"]
    )["fit_success"]

    probability = grouped.mean()
    count = grouped.count()

    # --------------------------------------------------------
    # CREATE 3D ARRAYS
    # --------------------------------------------------------

    probs = np.zeros((
        N_GAMMA1_BINS,
        N_GAMMA2_BINS,
        N_INTENSITY_BINS
    ))

    filled = np.zeros_like(probs, dtype=bool)

    for (i, j, k), p in probability.items():

        if np.isnan(i) or np.isnan(j) or np.isnan(k):
            continue

        i = int(i)
        j = int(j)
        k = int(k)

        probs[i, j, k] = p
        filled[i, j, k] = True

    # --------------------------------------------------------
    # COLORS
    # --------------------------------------------------------

    cmap = cm.get_cmap("RdYlGn")

    norm = Normalize(vmin=0, vmax=1)

    colors = cmap(norm(probs))

    # transparency
    colors[..., -1] = filled * 0.8

    # --------------------------------------------------------
    # PLOT
    # --------------------------------------------------------

    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111, projection='3d')

    ax.voxels(
        filled,
        facecolors=colors,
        edgecolor='k'
    )

    ax.set_xlabel("Gamma 1 Bin")
    ax.set_ylabel("Gamma 2 Bin")
    ax.set_zlabel("Rel. Intensity Bin")

    title = (
        f"Peak Distance Range: "
        f"{dmin:.4f} ≤ Δx < {dmax:.4f}"
    )
    ax.set_title(title)

    # --------------------------------------------------------
    # COLORBAR
    # --------------------------------------------------------

    mappable = cm.ScalarMappable(
        norm=norm,
        cmap=cmap
    )

    mappable.set_array([])

    cbar = plt.colorbar(
        mappable,
        ax=ax,
        shrink=0.7
    )

    cbar.set_label("Fit Success Probability")

    plt.tight_layout()
    plt.show()

# ============================================================
# MAIN LOOP OVER DISTANCE BINS
# ============================================================

for bin_id in range(N_DISTANCE_BINS):

    subdf = df[df["distance_bin"] == bin_id]

    dmin = distance_bins[bin_id]
    dmax = distance_bins[bin_id + 1]

    print(f"\nDistance Bin {bin_id}")
    print(f"Samples: {len(subdf)}")

    # --------------------------------------------------------
    # RAW SCATTER
    # --------------------------------------------------------

    plot_raw_scatter(subdf, bin_id, dmin, dmax)

    # --------------------------------------------------------
    # PROBABILITY VOLUME
    # --------------------------------------------------------

    plot_probability_volume(subdf, bin_id, dmin, dmax)
```

# Symbolic Regression

```python
import pandas as pd
import numpy as np

df = pd.read_csv("curve_summary.csv")

X = df[["relative_intensity", "gamma_1", "gamma_2", "peak_distance"]].values
y = df["fit_success"].values
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
from pysr import PySRRegressor

model = PySRRegressor(
    niterations=1000,

    # Keep it simple first
    binary_operators=["+", "-", "*", "/"],
    unary_operators=[],

    # Encourage simplicity
    model_selection="best",
    parsimony=1e-3,

    loss="loss(x, y) = (x - y)^2",

    # Limit complexity (VERY important)
    maxsize=20,
)

model.fit(X, y)
```

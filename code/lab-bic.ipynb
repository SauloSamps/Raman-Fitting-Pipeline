```python
from RamanModel import generateCurve
from AdaptiveKnn import *
from HelperFunctions import *
import matplotlib.pyplot as plt
```

# Example 1 - 5 peaks

```python
parameters = {
    "center_range": (0.1, 0.9),
    "amplitude_range": (0.1, 1),
    "gamma_range": (0.001, 0.05),
    "x_min": 0,
    "x_max": 1
}

x = np.linspace(0, 1, 1876)
x, y, params = generateCurve(parameters, 5, len(x), noise_sigma=0.05)
plt.plot(x, y)
```

```python
from AdaptiveKnn import *
import matplotlib.ticker as ticker

bic_values =[]
i_values = []

for i in range (1, 11):
    neighbors = knnFit_persistent(
            samples=1000,
            base_parameters=parameters,
            target_x=x,
            target_y=y,
            num_peaks=i,
            k=5,
            rounds=3,
            eps=0.15
        )
        
    # 2. Refine the parameters using non-linear least squares optimization
    best_fit_result = refine_knn_with_lmfit(neighbors, x, y)
    bic_values.append(float(best_fit_result.bic))
    i_values.append(i)
    

fig, ax = plt.subplots()
ax.plot(i_values, bic_values)
ax.xaxis.set_major_locator(ticker.MaxNLocator(integer=True))

plt.show()
```

```python
for idx, value in enumerate(bic_values):
    print(idx+1, value)
```
